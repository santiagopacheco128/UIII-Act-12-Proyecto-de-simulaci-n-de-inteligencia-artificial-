```python
# ============================================
# Sistema Inteligente de Guardias 🛡️
# Programación Orientada a Objetos
# ============================================

class Guardia:

    def __init__(self, empresa, puesto, equipo, turno, protocolos, capacitacion):
        self.empresa = empresa
        self.puesto = puesto
        self.equipo = equipo
        self.turno = turno
        self.protocolos = protocolos
        self.capacitacion = capacitacion


# Lista para guardar guardias
lista_guardias = []


# FUNCIONES
def agregar_guardia():

    print("\n📌 AGREGAR GUARDIA")

    empresa = input("Empresa de seguridad: ")
    puesto = input("Puesto de vigilancia: ")
    equipo = input("Equipo de comunicación: ")
    turno = input("Turno: ")
    protocolos = input("Protocolos de acceso: ")
    capacitacion = input("Capacitación: ")

    guardia = Guardia(
        empresa,
        puesto,
        equipo,
        turno,
        protocolos,
        capacitacion
    )

    lista_guardias.append(guardia)

    print("✅ Guardia agregado correctamente")


def listar_guardias():

    print("\n📋 LISTA DE GUARDIAS")

    if len(lista_guardias) == 0:
        print("No hay guardias registrados")

    else:
        # Adjusting column widths for better alignment
        print("--------------------------------------------------------------------------------------------------")
        print(f"{'EMPRESA':15} {'PUESTO':20} {'EQUIPO':18} {'TURNO':10} {'PROTOCOLOS':15} {'CAPACITACIÓN':15}")
        print("--------------------------------------------------------------------------------------------------")

        for guardia in lista_guardias:

            print(f"{guardia.empresa:15} {guardia.puesto:20} {guardia.equipo:18} {guardia.turno:10} {guardia.protocolos:15} {guardia.capacitacion:15}")

        print("--------------------------------------------------------------------------------------------------")


def actualizar_guardia():

    print("\n✏️ ACTUALIZAR GUARDIA")

    empresa_buscar = input("Ingresa la empresa del guardia: ")

    for guardia in lista_guardias:

        if guardia.empresa == empresa_buscar:

            print("✅ Guardia encontrado")

            guardia.puesto = input("Nuevo puesto de vigilancia: ")
            guardia.equipo = input("Nuevo equipo de comunicación: ")
            guardia.turno = input("Nuevo turno: ")
            guardia.protocolos = input("Nuevos protocolos de acceso: ")
            guardia.capacitacion = input("Nueva capacitación: ")

            print("✅ Información actualizada")
            return

    print("❌ Guardia no encontrado")


def borrar_guardia():

    print("\n🗑️ BORRAR GUARDIA")

    empresa_buscar = input("Ingresa la empresa del guardia: ")

    for guardia in lista_guardias:

        if guardia.empresa == empresa_buscar:

            lista_guardias.remove(guardia)

            print("✅ Guardia eliminado correctamente")
            return

    print("❌ Guardia no encontrado")


# MENÚ PRINCIPAL
while True:

    print("\n================================")
    print("🛡️ SISTEMA INTELIGENTE DE GUARDIAS")
    print("================================")
    print("1. Agregar guardia")
    print("2. Listar guardias")
    print("3. Actualizar guardia")
    print("4. Borrar guardia")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        agregar_guardia()

    elif opcion == "2":
        listar_guardias()

    elif opcion == "3":
        actualizar_guardia()

    elif opcion == "4":
        borrar_guardia()

    elif opcion == "5":
        print("👋 Saliendo del sistema...")
        print("Programa realizado por Santiago Ernesto Pacheco Holguin")
        break

    else:
        print("❌ Opción inválida")
```
